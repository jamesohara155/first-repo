```python
##bins should be equal sizes, round numbers, bins also come from realistic reasons
```

```python
bins_we_made = [0,100000,200000,300000,400000,500000,600000,700000,800000,900000,1000000, float('inf')] ## create bins, float('inf') means setting last bin limit at infinity
```

```python
pd.cut(salaries['EMPLOYEE ANNUAL BASE SALARY'],
       bins=bins_we_made,
       right=False).value_counts().sort_index().to_frame() #apply bins to salaries, count how many people in each bin, sort groups in size order, apply to dataframe
```

```python
salaries['JOB CATEGORY RECODED'] = salaries['JOB CATEGORY'].case_when(
  [
    (salaries['JOB CATEGORY'].str.contains(
      'coach', case=False, na=False),
    'Coach'

    )
,

    (salaries['JOB CATEGORY'].str.contains(
      'professor', case=False, na=False),
    'Professor'

   )
  ]
)
```

```python
salaries['JOB CATEGORY RECODED B'] = salaries['JOB CATEGORY'].case_when(
  [(salaries['JOB CATEGORY'].isin(['Administrative Assistant', "IT Support", 'Facilities']),
   'Back Office'
  )
]
  )
```
